# 文本清理演示

本 notebook 演示如何使用 Python 进行文本清理，包括去除噪声、标准化文本格式等操作。

## 1. 导入必要的库

```python
import re
import pandas as pd
```

## 2. 读取数据

我们将使用之前生成的 cleaned_dataset.json 文件进行演示。

```python
import json

# 读取 cleaned_dataset.json 文件
with open('cleaned_dataset.json', 'r', encoding='utf-8') as f:
    data = json.load(f)

# 转换为 DataFrame
df = pd.DataFrame(data)

print('数据形状:', df.shape)
print('数据前5行:')
df.head()
```

## 3. 文本清理函数

我们将创建一系列函数来清理文本。

```python
def clean_text(text):
    """清理文本的函数"""
    # 1. 去除首尾空白字符
    text = text.strip()
    
    # 2. 统一转换为小写
    text = text.lower()
    
    # 3. 去除多余的标点符号
    text = re.sub(r'[!]+', '!', text)
    text = re.sub(r'[?]+', '?', text)
    text = re.sub(r'[.]+', '.', text)
    
    # 4. 去除多余的空格
    text = re.sub(r'\s+', ' ', text)
    
    # 5. 去除特殊字符（可选）
    # text = re.sub(r'[^a-zA-Z0-9\s.!?]', '', text)
    
    return text

def remove_stopwords(text):
    """去除停用词的函数"""
    # 简单的中文停用词列表
    stopwords = ['的', '了', '在', '是', '我', '有', '和', '就', '不', '人', '都', '一', '一个', '上', '也', '很', '到', '说', '要', '去', '你', '会', '着', '没有', '看', '好', '自己', '这']
    
    words = text.split()
    filtered_words = [word for word in words if word not in stopwords]
    
    return ' '.join(filtered_words)

def text_preprocess(text):
    """文本预处理主函数"""
    text = clean_text(text)
    text = remove_stopwords(text)
    return text
```

## 4. 应用文本清理函数

```python
# 应用文本清理函数
df['processed_prompt'] = df['raw_prompt'].apply(text_preprocess)

# 查看清理前后的对比
print('清理前后的对比:')
for i in range(5):
    print(f'原始文本: {df['raw_prompt'][i]}')
    print(f'清理后: {df['processed_prompt'][i]}')
```

## 5. 文本特征提取

```python
# 提取文本长度特征
df['raw_length'] = df['raw_prompt'].apply(len)
df['processed_length'] = df['processed_prompt'].apply(len)

# 提取词数特征
df['raw_word_count'] = df['raw_prompt'].apply(lambda x: len(x.split()))
df['processed_word_count'] = df['processed_prompt'].apply(lambda x: len(x.split()))

# 查看特征提取结果
print('特征提取结果:')
df[['raw_prompt', 'processed_prompt', 'raw_length', 'processed_length', 'raw_word_count', 'processed_word_count']].head()
```

## 6. 数据统计与分析

```python
# 统计清理前后的长度变化
print('清理前后的长度统计:')
print('原始文本长度:', df['raw_length'].describe())
print('清理后文本长度:', df['processed_length'].describe())

# 统计清理前后的词数变化
print('清理前后的词数统计:')
print('原始文本词数:', df['raw_word_count'].describe())
print('清理后文本词数:', df['processed_word_count'].describe())

# 统计标签分布
print('标签分布:')
print(df['label'].value_counts())
```

## 7. 导出清理后的数据

```python
# 导出清理后的数据
df.to_csv('cleaned_text_data.csv', index=False, encoding='utf-8')
print('清理后的数据已导出为 cleaned_text_data.csv')
```
