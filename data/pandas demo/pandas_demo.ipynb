# Pandas DataFrame 小练习

本练习旨在熟悉 Pandas 库的基本操作，包括数据读取、清洗、转换和分析。

## 1. 导入必要的库

```python
import pandas as pd
import numpy as np
```

## 2. 读取数据

我们将使用目录中的 CSV 文件(注入攻击.csv、良性prompt.csv)进行练习。

```python
# 读取良性prompt.csv文件
benign_df = pd.read_csv('良性prompt.csv')

# 读取注入攻击.csv文件
attack_df = pd.read_csv('注入攻击.csv')

# 查看良性prompt数据的基本信息
print('良性prompt数据形状:', benign_df.shape)
print('良性prompt数据列:', benign_df.columns)
print('良性prompt数据前5行:')
benign_df.head()
```

```python
# 查看注入攻击数据的基本信息
print('注入攻击数据形状:', attack_df.shape)
print('注入攻击数据列:', attack_df.columns)
print('注入攻击数据前5行:')
attack_df.head()
```

## 3. 数据清洗

### 3.1 处理缺失值

```python
# 检查良性prompt数据中的缺失值
print('良性prompt数据缺失值统计:')
print(benign_df.isnull().sum())

# 检查注入攻击数据中的缺失值
print('注入攻击数据缺失值统计:')
print(attack_df.isnull().sum())

# 处理缺失值（如果有）
benign_df = benign_df.dropna()
attack_df = attack_df.dropna()

print('处理后的数据形状:')
print('良性prompt:', benign_df.shape)
print('注入攻击:', attack_df.shape)
```

### 3.2 数据类型转换

```python
# 查看数据类型
print('良性prompt数据类型:')
print(benign_df.dtypes)

print('注入攻击数据类型:')
print(attack_df.dtypes)
```

## 4. 数据合并与标签处理

我们将两个数据集合并，并添加标签列。

```python
# 为数据集添加标签
benign_df['label'] = 0  # 0表示良性
attack_df['label'] = 1  # 1表示攻击

# 合并数据集
combined_df = pd.concat([benign_df, attack_df], ignore_index=True)

# 随机打乱数据（神奇的42登场了）
combined_df = combined_df.sample(frac=1, random_state=42).reset_index(drop=True)

print('合并后的数据形状:', combined_df.shape)
print('合并后的数据前5行:')
combined_df.head()
```

## 5. 数据统计与分析

```python
# 统计标签分布
print('标签分布:')
print(combined_df['label'].value_counts())

# 计算prompt长度
combined_df['prompt_length'] = combined_df.iloc[:, 0].apply(len)

# 统计长度信息
print('Prompt长度统计:')
print(combined_df['prompt_length'].describe())
```

## 6. 数据导出

将处理后的数据导出为新的CSV文件。

```python
# 导出为CSV文件
combined_df.to_csv('processed_prompts.csv', index=False, encoding='utf-8')
print('数据已导出为 processed_prompts.csv')
```
